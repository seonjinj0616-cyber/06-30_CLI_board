```python
num_list = []
title_list = []
detail_list = []
i = 1

print("== CLI 게시판 실행==")
while True:

  명령어 = input("명령어 ) ")#.strip() 을 붙이면 양쪽 띄어쓰기도 명령어 인식 가능

  if 명령어 == "exit":
    print("== CLI 게시판 종료==")
    break

  elif 명령어 == "article write":

    제목 = input("제목: ")
    내용 = input("내용: ")
    print("{}번 글이 생성 되었습니다.".format(i))
    num_list += [i]
    title_list += [제목]
    detail_list += [내용]
    i += 1

  elif 명령어 == "article list":
    print("============================")
    print("번호   /   제목   /   내용")
    for i in range(len(num_list)-1,-1,-1):
      print(" {}    /    {}    /    {}".format(num_list[i],title_list[i],detail_list[i]))
    print("============================")

  else:
    print("지원하지 않는 기능입니다.")
```

```python
print("== CLI 게시판 실행==")

last_article_id = 0
articles = []

while True:

  cmd = input("명령어 ) ").strip() #붙이면 양쪽 띄어쓰기도 명령어 인식 가능

  if cmd == "exit":
    break

  elif cmd == "article write":
    last_article_id += 1
    id = last_article_id
    title = input("제목: ")
    body = input("내용: ")
    article = {"id" : id, "title" : title, "body" : body}  # dict로 1개 변수로 조립
    articles.append(article)
    print("{}번 글이 생성 되었습니다.".format(last_article_id))

  elif cmd == "article list":
    if not articles:
      print("게시글이 없습니다")
    else:
      print("============================")
      print("번호   /   제목   /   내용")
      for a in reversed(articles):
        print(" {}    /    {}    /    {}".format(a["id"],a["title"],a["body"]))
      print("============================")

  else:
    print("지원하지 않는 기능입니다.")

print("== CLI 게시판 종료==")
```
