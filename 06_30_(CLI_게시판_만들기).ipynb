```python
from datetime import datetime

# ===== 공통 함수 (중복 제거) =====

# 함수의 변수들은 미리 써도 상관없어
# 왜? 어짜피 함수 실행할 때 다시 올라와서 실행하니까
# 그러니까 아래(무한 루프)에서 변수 잡아놓고
# 사용할 변수를 함수에다가 넣는건데
# 미리 아래에서 잡아놓은 변수를 함수에 넣어서 사용

def now():
    # 현재 시각을 "YYYY-MM-DD HH:MM:SS" 문자열로
    return datetime.now().strftime("%Y-%m-%d %H:%M:%S")

    # datetime.now().strftime("%Y-%m-%d %H:%M:%S") 값의 변수를 함수로 묶었어 (결국 변수)

def parse_id(cmd):
    # "article delete 3" -> 3 을 뽑아준다. 잘못된 명령이면 None

    # 함수 안에 있는 cmd는 처음 input으로 넣은 cmd
    cmdBits = cmd.split(" ")
    # cmd 쪼개서 3개 list만든거 comBits 넣고
    if len(cmdBits) < 3:
        print("명령어를 확인해주세요(길이 모자람)")
        return None

    # 똑같이 작성했는데, 굳이 왜 return None을 넣었을까?
    # 아래 코드에서 None이 반환되었을 때, 조건을 사용하기 위해서
    # return으로 None을 나온다는 뜻은 내 조건에 맞지 않는다는거 = 사용이 X
    if not cmdBits[2].isdigit():
        print("명령어를 확인해주세요(번호대신 str 입력)")
        return None
    # 위랑 똑같이 조건이 맞지 않으면, 위에 말 출력하고
    # None 값 리턴해서 조건이 x = 사용 x
    return int(cmdBits[2])
    # parse_id 함수 반환값으로 int(cmdBits[2])
    # 왜? cmd의 뒷 숫자를 원하는데 이 함수는 숫자를 선택하기 전에 숫자 형식이 맞는지 아닌지
    # 확인해주는 함수
    # 그래서 이 함수는 결국 cmd를 넣어주면 숫자인지 아닌지 확인해주고 맞으면 숫자면 반환해줘
    # 숫자를 반환한다 = 검사할 준비가 되어있다. = 아직 검사 x = 내가 찾을 숫자만 반환

def find_article(articles, article_id):
    # id가 맞는 글을 돌려준다. 없으면 None (found 플래그가 필요 없어짐)

    # 숫자 찾았으니 이제 대조해봐야겠지
    # 매개변수로는 전체 게시글 들어있는(articles)
    # 찾으려는 게시글 id(article_id) == parse_id(cmd)
    # 왜 두개를 넣었나?
    # 찾는 행위(코드)가 필요한게 전체랑 찾는거 2개니까
    # 전의 코드도 2개를 비교했고
    for article in articles:
        if article["id"] == article_id:
            return article
    return None

    # 그래서 나온게 article
    # article은 내가 찾고자 하는 게시글을 담고있는 딕셔너리
    # 여기서도 return None이 나온이유는 모르겠네
    # 아래 코드를 보니 None 유무로 이 함수를 통해 내가 원하는 걸 찾았고, 못 찾았고로 2분법으로 나타내기 위해서 사용했네
    # 그래서 조건문으로 None 이면 작동 못하게 하는 식으로

    # 그래서 이 함수는 전체 게시글(전체 list) 에서 내가 찾는거(아무 변수나 넣어, 어짜피 parse_id(cmd)만 맞추면 되니까)
    # parse_id(cmd) return으로 int(cmdBits[2]) 나오니까 즉 내가 찾는 게시물 id를 리턴
    # 를 넣어서 내가 찾는게 있는지 없는지 확인
    # 있으면 게시물(딕셔너리) 주고 없으면 None 값 반환

def format_date(regDate):
    # 오늘 쓴 글이면 시:분:초, 이전 글이면 연-월-일 로 보여준다

    # 게시물 시간 or 날짜 나오는거 정리한거 같은데
    # 여긴 regDate를 넣었네, 왜?
    # 그럼 regDate는 뭔데?
    # 아래 루프에서 regDate = now()함수야 그러니까 = datetime.now().strftime("%Y-%m-%d %H:%M:%S") = 현재 시간을 넣은거지
    # 함수 안에서 datePart, timePart 변수를 만들려고
    # 왜? 현재 시간과 작성 시간을 비교할라고
    # datepart = 작성 날짜, timepart = 작성 시각
    # 그래서 날짜랑 오늘 날짜 같으면 시간 리턴하고 다르면 날짜 리턴하게 짜셨네
    # 그럼 이 함수 쓰면, 작성날짜 현재 시간 비교해서 날짜, 시간 골라서 리턴해주네

    today = now().split(" ")[0]        # "2026-06-30"
    # 현재 시간 쪼개서 [날짜, 시간] 중 날짜만 today에 담고
    datePart = regDate.split(" ")[0]   # 글의 날짜
    # 잘라서 앞부분(날짜)
    timePart = regDate.split(" ")[1]   # 글의 시각
    # 잘라서 뒷부분 (시각)
    if datePart == today:    # 작성 날짜 == 오늘 날짜
        return timePart      # 맞으면 시간
    else:
        return datePart      # 다르면 날짜


# ===== 메인 =====

print("== CLI 게시판 실행 ==")

last_article_id = 1
articles = [
    {"id": 1, "regDate": "2025-12-12 12:12:12", "updateDate": "2025-12-12 12:12:12", "title": "제목1", "body": "내용1"},
]

while True:
    cmd = input("명령어 ) ").strip()

    if cmd == "exit":
        break
# 여기까지는 똑같아
# 시작할때 cmd 쓰는 방식은 똑같으니까, 뭐 종료도 똑같고
# 문제는 위에 만든 함수로 어떻게 줄이냐 인데
    elif cmd == "article write":
        last_article_id += 1
        title = input("제목 : ")
        body = input("내용 : ")
        article = {
            "id": last_article_id,
            "regDate": now(),            # 그냥 now() 함수로 시간 변수 대신 넣었고
            "updateDate": now(),         # 그냥 now() 함수로 시간 변수 대신 넣었고
            "title": title,
            "body": body,
        }                              # 딕셔너리 깔끔하게 한줄씩 정리하셨네
        articles.append(article)
        print("{}번 글이 생성되었습니다".format(last_article_id))

    elif cmd == "article list":
        if not articles:
            print("게시글이 없습니다")
        else:
            print("=========================================")
            print("   번호      /     제목       /     내용   /     작성일   ")
            for a in reversed(articles):
                print("   {}      /     {}       /     {}          /     {}   ".format(
                    a["id"], a["title"], a["body"], format_date(a["regDate"])))
            print("=========================================")
# article list는 중복되는게 없으니 그냥 그대로 사용하셨고

    elif cmd.startswith("article delete"):
        deletedId = parse_id(cmd)           # 여기서 parse_id(cmd) 함수를 사용했고, 함수 대신에 변수로
        if deletedId is None:               # 변수 parse_id(cmd) = None => 글자 조건이 X
            continue                        # continue로 올라가면 반복문까지 올라가니까
                                            # 다시 cmd 작성하겠네

        article = find_article(articles, deletedId) # find_article 함수 이용해서 내가 원하는 함수 찾아
      # articles는 전체 게시글이고, deletedid = parse_id(cmd) 니까 이건 return으로 내가 찾고자하는 게시물의 id를 리턴
      # parse_id(cmd) = int(cmdBits[2])
      # 그리고 find_article 함수에 의해 리턴되는 값은 "내가 찾고자 하는 딕셔너리" / 못 찾으면 None 리턴
      # 를 article 변수에 담으셨고
        if article is None:                   # "내가 찾고자 하는 딕셔너리"가 없으면(None 이면)
            print("{}번 글은 없습니다".format(deletedId))  # 없다고 하고
            continue                                     # 다시 cmd 작성으로 돌려

        articles.remove(article)                  # 만약 조건문이 False면, 이 문장이 실행되고 내가 원하는 글 삭제
        print("{}번 글이 삭제되었습니다".format(deletedId))

    elif cmd.startswith("article edit"):    # 수정부분
        editedId = parse_id(cmd)          # 동일하게 내가 찾고자하는 id 작성 오류 검사, 내가 원하는 id 숫자 return
        if editedId is None:             # None 이라는건 조건이 맞지 x = 못써
            continue                        # 다시 cmd 써

        article = find_article(articles, editedId)   # 맞으면 if 조건문 뒤의 다음 코드가 실행
        if article is None:                          # find_article을 이용하여 내가 원하는 id가 전체에 없으면
            print("{}번 글은 없습니다".format(editedId)) # 없어
            continue  # 다시 cmd 써

        print("기존 제목 : {}".format(article["title"]))
        print("기존 내용 : {}".format(article["body"]))  # 전 내용이 뭔지 알려주고
        # 여기서부터는 중복되는게 없으니 그대로 작성
        article["title"] = input("새 제목 : ")          # 고친 문장
        article["body"] = input("새 내용 : ")           # 고친내용
        article["updateDate"] = now()
        print("{}번 글이 수정되었습니다".format(editedId))

    elif cmd.startswith("article detail"):        # 상세정보
        detailId = parse_id(cmd)            # cmd 규격 먼저 검사하고
        if detailId is None:                # 안 맞아
            continue                        # cmd 다시 써

        article = find_article(articles, detailId)   # 맞으면 detailId = 내가 찾는 id야, 전체에서 비교해봐
        if article is None:                          # 없어?
            print("{}번 글은 없습니다".format(detailId)) # 그럼 없어
            continue                                 # cmd 다시 써
        # 만약 있으면 그대로 내가 찾는 id에 관한 정보가 출력되도록 작성
        print("번호 : {}".format(article["id"]))
        print("작성 날짜 : {}".format(article["regDate"]))
        print("수정 날짜 : {}".format(article["updateDate"]))
        print("제목 : {}".format(article["title"]))
        print("내용 : {}".format(article["body"]))

    else:
        print("지원하지 않는 기능입니다")

print("== CLI 게시판 종료 ==")
```
