```python
print("== CLI 게시판 실행 ==")

last_article_id = 0
articles = [] # 게시글 저장소

while True:
    cmd = input("명령어 ) ").strip().split(" ")

    if cmd[0] == "exit":
      break

    elif cmd[1] == "write":
      last_article_id += 1
      id = last_article_id
      title = input("제목 : ")
      body = input("내용 : ")
      article = {"id" : id, "title" : title, "body" : body}
      articles.append(article)
      print("{}번 글이 생성되었습니다".format(last_article_id))

    elif cmd[1] == "list":
      if not articles:
        print("게시글이 없습니다")
      else:
        print("=========================================")
        print("   번호      /     제목       /     내용   ")
        for a in reversed(articles):
          print("   {}      /     {}       /     {}   ".format(a["id"], a["title"], a["body"]))
        print("=========================================")

    elif cmd[1] == "delete":
      if int(cmd[2])

      for d in articles:
        if int(cmd[2]) == d["id"]:
          d.clear()
          print("{}번 글이 삭제 되었습니다.".format(cmd[2]))
          print(articles)




    else:
      print("지원하지 않는 기능입니다")

print("== CLI 게시판 종료 ==")
```
