```python
while True:

  명령어 = input("명령어?")

  if 명령어 == "exit":
    print("== CLI 게시판 종료==")
    break

  elif 명령어 == "ㅇ":
    print("== CLI 게시판 실행==")

    제목 = input("제목?")
    내용 = input("내용?")
    i = 1

    # print("명령어) {}".format(명령어))
    print("제목: {}".format(제목))
    print("내용: {}".format(내용))
    print("{}번 글이 생성 되었습니다.".format(i))
    i += 1
```
