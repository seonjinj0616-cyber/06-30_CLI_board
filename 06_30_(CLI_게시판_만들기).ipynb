```python
num_list = []
title_list = []
detail_list = []
i = 1

while True:

  명령어 = input("명령어 ) ")

  if 명령어 == "exit":
    print("== CLI 게시판 종료==")
    break

  elif 명령어 == "article write":
    print("== CLI 게시판 실행==")

    제목 = input("제목: ")
    내용 = input("내용: ")

    print("제목: {}".format(제목))
    print("내용: {}".format(내용))
    print("{}번 글이 생성 되었습니다.".format(i))
    num_list += [i]
    title_list += [제목]
    detail_list += [내용]
    i += 1


  elif 명령어 == "article list":
    print("============================")
    print("번호   /   제목   /   내용")
    for i in range(len(num_list)-1,-1,-1):
      print(" {}    /    {}    /    {}".format(num_list[i],title_list[i],detail_list[i]))
    print("============================")

  else:
    print("지원하지 않는 기능입니다.")
```
